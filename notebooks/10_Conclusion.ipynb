## Core Business Answers

### 1. What is Customer Churn?
Customer churn is the percentage of subscribers who terminate their service contracts during a given timeframe. In the IBM Telco dataset, the baseline churn rate is **26.54%** (1,869 churners out of 7,043 customers).

### 2. Why is Predicting Churn Critical?
- Acquiring a new telecom customer costs \$300 to \$500, compared to \$50-\$100 to retain one.
- Churn directly erodes monthly recurring revenue (MRR) and customer lifetime value (CLV).
- Identifying at-risk customers early enables targeted, cost-effective retention campaigns.

### 3. Which Customers Are Most Likely to Leave?
1. **Contract Type:** Month-to-month subscribers have a **42.7% churn rate**, compared to **11.3%** for 1-year and **2.8%** for 2-year contracts.
2. **Tenure:** Customers with tenure under 12 months have over **47% churn risk**.
3. **Internet Service:** Fiber optic subscribers churn at **41.9%**, driven by high monthly fees.
4. **Payment Method:** Electronic check customers churn at **45.3%**, compared to <17% for automatic bank/card payments.
5. **Add-on Services:** Customers lacking `OnlineSecurity` or `TechSupport` churn at over **41%**.

### 4. Which Factors Influence Churn the Most?
Based on feature importances and logistic regression coefficients:
- **Strongest Positive Drivers (Increase Churn):** Month-to-month contracts, Fiber optic service, Electronic check payments, High monthly charges.
- **Strongest Negative Drivers (Reduce Churn):** Long tenure, Two-year contracts, Online Security, Tech Support, Automatic payment methods.

### 5. Which Model Was Selected and Why?
- **Selected Model:** **Logistic Regression / Tuned Random Forest**.
- **Justification:** Logistic Regression achieved top test set **F1 Score (0.604)**, top **ROC-AUC (0.842)**, rapid inference, and linear interpretability. Tuned Random Forest achieved top **Accuracy (81.26%)** and **Precision (68.8%)**.
- Both models generalize reliably across 5-fold cross-validation without overfitting.

### 6. How Reliable Are the Predictions?
- Evaluated on an independent, stratified 20% test set (1,409 customers) with zero data leakage.
- Verified across 5-fold cross-validation with low variance ($F_1$ std < 0.02).

### 7. Strategic Retention Playbook:
| Risk Tier | Probability Range | Action Plan |
|---|---|---|
| **Low Risk** | 0.00 - 0.29 | Maintain standard experience. Do not discount. Eligible for premium upselling. |
| **Medium Risk** | 0.30 - 0.59 | Automated satisfaction survey, connection check-in, offer 10% discount for 1-year contract migration. |
| **High Risk** | 0.60 - 1.00 | Dedicated retention specialist outreach within 24 hours, tech support review, custom pricing incentive for 2-year commitment. |

### 8. Operational Threshold Recommendation:
Adjusting the operational cutoff from the default **0.50** to **0.35** raises churner recall from **55.9% to 70.6%**, capturing an additional 55 at-risk accounts per 1,400 customers.
